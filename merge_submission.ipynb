```python
import pandas as pd 


sub_1="/Users/muhsenalzzaqry/Desktop/CAFA_6_BASE_LINE/models/t5_tax/submission.tsv" 

sub_2="/Users/muhsenalzzaqry/Desktop/CAFA_6_BASE_LINE/models/submission.tsv"


submission1 = pd.read_csv( sub_1,
                          sep='\t', header=None, names=['Id', 'GO term', 'Confidence']) 

submission2 = pd.read_csv( sub_2,
                          sep='\t', header=None, names=['Id', 'GO term', 'Confidence'])  

merged = submission1.merge(submission2,
                           on=['Id', 'GO term'],
                           how='outer',
                           suffixes=('1', '2')) 


merged['Confidence1'] = merged['Confidence1'].fillna(0)
merged['Confidence2'] = merged['Confidence2'].fillna(0)



W1 = 0.5 
W2 = 0.5 

merged['Confidence'] = (W1 * merged['Confidence1']) + (W2 * merged['Confidence2'])

final_submission = merged[['Id', 'GO term', 'Confidence']]
```
